```python
import pandas as pd
import numpy as np

from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix
)
```

```python
X_train = pd.read_csv("X_train_processed.csv")
X_test = pd.read_csv("X_test_processed.csv")

y_train = pd.read_csv("y_train.csv").values.ravel()
y_test = pd.read_csv("y_test.csv").values.ravel()

X_train.shape, X_test.shape
```

```python
X_train.head()
np.unique(y_train)
```

```python
log_reg = LogisticRegression(
    max_iter=1000,
    multi_class="auto",
    random_state=42
)
```

```python
log_reg.fit(X_train, y_train)
```

```python
log_reg.coef_.shape
```

```python
y_pred_lr = log_reg.predict(X_test)

print("Accuracy:", accuracy_score(y_test, y_pred_lr))
print("\nClassification Report:\n", classification_report(y_test, y_pred_lr))
```

```python
confusion_matrix(y_test, y_pred_lr)
```

```python
dt = DecisionTreeClassifier(
    max_depth=5,
    random_state=42
)
```

```python
dt.fit(X_train, y_train)
```

```python
dt.get_depth()
```

```python
y_pred_dt = dt.predict(X_test)

print("Accuracy:", accuracy_score(y_test, y_pred_dt))
print("\nClassification Report:\n", classification_report(y_test, y_pred_dt))
```

```python
baseline_results = pd.DataFrame({
    "Model": ["Logistic Regression", "Decision Tree"],
    "Accuracy": [
        accuracy_score(y_test, y_pred_lr),
        accuracy_score(y_test, y_pred_dt)
    ]
})

baseline_results
```

```python
# Check class balance
pd.Series(y_train).value_counts(normalize=True)

# Check prediction distribution
pd.Series(y_pred_lr).value_counts()
pd.Series(y_pred_dt).value_counts()
```

```python
log_reg_bal = LogisticRegression(
    max_iter=1000,
    class_weight="balanced",
    multi_class="auto",
    random_state=42
)
```

```python
log_reg_bal
```

```python
log_reg_bal.fit(X_train, y_train)
```

```python
log_reg_bal.coef_.shape
```

```python
y_pred_lr_bal = log_reg_bal.predict(X_test)

print("Accuracy:", accuracy_score(y_test, y_pred_lr_bal))
print("\nClassification Report:\n", classification_report(y_test, y_pred_lr_bal))
```

```python
pd.Series(y_pred_lr_bal).value_counts()
```

```python
dt_bal = DecisionTreeClassifier(
    max_depth=5,
    class_weight="balanced",
    random_state=42
)
```

```python
dt_bal.fit(X_train, y_train)
```

```python
dt_bal.get_depth()
```

```python
y_pred_dt_bal = dt_bal.predict(X_test)

print("Accuracy:", accuracy_score(y_test, y_pred_dt_bal))
print("\nClassification Report:\n", classification_report(y_test, y_pred_dt_bal))
```

```python
baseline_comparison = pd.DataFrame({
    "Model": [
        "Logistic Regression",
        "Logistic Regression (Balanced)",
        "Decision Tree",
        "Decision Tree (Balanced)"
    ],
    "Accuracy": [
        accuracy_score(y_test, y_pred_lr),
        accuracy_score(y_test, y_pred_lr_bal),
        accuracy_score(y_test, y_pred_dt),
        accuracy_score(y_test, y_pred_dt_bal)
    ]
})

baseline_comparison
```

```python
from sklearn.metrics import recall_score

print("Recall (Malnourished):")
print("LR:", recall_score(y_test, y_pred_lr, labels=["Malnourished"], average=None))
print("LR Balanced:", recall_score(y_test, y_pred_lr_bal, labels=["Malnourished"], average=None))
print("DT:", recall_score(y_test, y_pred_dt, labels=["Malnourished"], average=None))
print("DT Balanced:", recall_score(y_test, y_pred_dt_bal, labels=["Malnourished"], average=None))
```
